```python
# Dynamic Task Assignment with Job Queue multi processing version main function
import multiprocessing
import os
from resource_change import *
from common_function import *

def process_sample(sample):
    sample_dir = '../../../../DikeDataset/malware/'
    save_dir = '../evaluation/perturbated_sample/instruction_change/' 
    
    samples = list_files_by_size(sample_dir)
    #print(samples)
    create_directory(save_dir)

    for sample in samples:
        if '.ipynb' in sample or '.pickle' in sample or '.txt' in sample or '.zip' in sample or '.' not in sample:
            continue
        
        print(sample)

        input_filepath = sample_dir+sample
        
        try:
            new_text = disassemble_and_modify(input_filepath, save_dir)

            if new_text is None:
                print(f"[+] Error : failed to make new_text section.")
            else:
                new_text = modify_headers(input_filepath, new_text)
                modify_section(input_filepath, new_text, save_dir)
        except pefile.PEFormatError:
            continue
            
def worker(input_queue):
    while True:
        sample = input_queue.get()
        if sample is None:
            break
        process_sample(sample)
        input_queue.task_done()

def main():
    sample_dir = '../../../../DikeDataset/malware/'
    save_dir = '../evaluation/perturbated_sample/instruction_change/' 
    
    samples = list_files_by_size(sample_dir)
    create_directory(save_dir)

    # 유효한 샘플만 필터링
    samples = [sample for sample in samples if not any(ext in sample for ext in ['.ipynb', '.pickle', '.txt', '.zip','.dll'])]

    # 작업 큐 생성
    input_queue = multiprocessing.JoinableQueue()

    # CPU 코어 수의 절반만 사용하도록 설정
    num_processes = max(1, multiprocessing.cpu_count() // 2)

    # 워커 프로세스 생성 및 시작
    processes = []
    for _ in range(num_processes):
        p = multiprocessing.Process(target=worker, args=(input_queue,))
        p.start()
        processes.append(p)

    # 작업 큐에 작업 추가
    for sample in samples:
        input_queue.put(sample)

    # 모든 작업이 완료되면 None을 넣어 워커 프로세스를 종료시킴
    input_queue.join()
    for _ in range(num_processes):
        input_queue.put(None)

    # 워커 프로세스 종료
    for p in processes:
        p.join()

    print("All tasks are completed.")

if __name__ == '__main__':
    main()
```

```python
import os
import multiprocessing
from instruction_change import *
from common_function import *

def process_sample(sample, root, save_dir):
    input_filepath = os.path.join(root, sample)
    output_filename = sample.replace('.exe', '_changing.exe')
    output_filepath = os.path.join(save_dir, output_filename)
    
    if os.path.isfile(output_filepath):
        return  # 이미 처리된 파일은 건너뜀

    try:
        new_text = disassemble_and_modify(input_filepath, save_dir)

        if new_text is None:
            print(f"[+] Error: Failed to create new_text section for {sample}.")
        else:
            new_text = modify_headers(input_filepath, new_text)
            modify_section(input_filepath, new_text, save_dir+'/')
    except pefile.PEFormatError:
        pass

def worker(input_queue):
    while True:
        task = input_queue.get()
        if task is None:
            break
        sample, root, save_dir = task
        process_sample(sample, root, save_dir)
        input_queue.task_done()

def main():
    sample_dir = '../sample/labeling/'
    save_dir_base = '../sample/perturbated_labling_sample/instruction_change/'
    
    # 작업 큐 생성
    input_queue = multiprocessing.JoinableQueue()

    # CPU 코어 수의 절반만 사용하도록 설정
    num_processes = max(1, multiprocessing.cpu_count() // 5)

    # 워커 프로세스 생성 및 시작
    processes = []
    for _ in range(num_processes):
        p = multiprocessing.Process(target=worker, args=(input_queue,))
        p.start()
        processes.append(p)

    for root, dirs, files in os.walk(sample_dir):
        if 'ok' in root.lower().split(os.sep):  # 'OK' 디렉토리를 건너뛰기
            continue
            
        if len(files) <= 10:  # 파일이 10개 이하인 디렉토리는 건너뛰기
            continue
        
        save_dir = os.path.join(save_dir_base, os.path.basename(root))
        create_directory(save_dir)
        
        for sample in files:
            if any(ext in sample for ext in ['.ipynb', '.pickle', '.txt', '.zip']) or '.' not in sample:
                continue
            
            input_queue.put((sample, root, save_dir))

    # 모든 작업이 완료되면 None을 넣어 워커 프로세스를 종료시킴
    input_queue.join()
    for _ in range(num_processes):
        input_queue.put(None)

    # 워커 프로세스 종료
    for p in processes:
        p.join()

    print("All tasks are completed.")

if __name__ == '__main__':
    main()
```

```python
import capstone

def disassemble(byte_code):
    # Capstone을 사용하여 x86 32비트 모드에서 디스어셈블리
    md = capstone.Cs(capstone.CS_ARCH_X86, capstone.CS_MODE_32)
    
    # 바이트코드 디스어셈블
    for instr in md.disasm(byte_code, 0x1000):
        print(f"{instr.mnemonic} {instr.op_str}")
#b'\x8d\x80\x83\x00\x00' b'\x05\x83\x00\x00\x00' 6 5 
# 바이트코드
#byte_code = b'-\x00\x00\xfc\xff'
#b'\x83\xee\xec' b'\x83\xc6\x14' 3 3 
mc_code = b'\x83\xc6\x14'
if '-' in str(mc_code):
    print('ini', str(mc_code))
print(len(mc_code))
# 디스어셈블리 호출
disassemble(mc_code)
```

```python
from keystone import *
from typing import List, Tuple

ks = Ks(KS_ARCH_X86, KS_MODE_32)

# 대체 가능한 명령어 목록 (원본 명령어에 따라 동작이 동일한 대체 명령어를 찾기 위해 필요)
instruction_map = {
    "mov ecx,": [
        "push {val}; pop ecx",
        # 추가 대체 명령어가 있으면 여기에 추가 가능
    ],
    # 다른 레지스터나 동작에 대해서도 추가 가능
}

def get_encoding(instruction: str) -> Tuple[List[int], int]:
    """Assemble an instruction and return its byte encoding and length."""
    encoding, count = ks.asm(instruction)
    return encoding, len(encoding)

def find_alternative(instruction: str) -> str:
    # 명령어 파싱 (단순화된 파싱: 레지스터와 값을 구분)
    parts = instruction.split()
    if len(parts) < 3:
        raise ValueError("Invalid instruction format")

    # 원본 명령어의 레지스터와 값 파싱
    opcode, reg, val = parts[0], parts[1].strip(","), parts[2]

    # 원본 명령어의 길이 계산
    original_encoding, original_length = get_encoding(instruction)

    # 대체 가능한 명령어 목록을 확인하여 동일한 동작을 하는 명령어 찾기
    alternatives = instruction_map.get(f"{opcode} {reg},", [])
    
    for alt in alternatives:
        # 대체 명령어의 길이를 확인하여 원본과 비교
        alt_instruction = alt.format(val=val)
        alt_encoding, alt_length = get_encoding(alt_instruction)
        
        if alt_length <= original_length:
            print(f"Original: {instruction} ({original_length} bytes)")
            print(f"Alternative: {alt_instruction} ({alt_length} bytes)")
            print(f"Bytes: {' '.join(format(x, '02x') for x in alt_encoding)}")
            return alt_instruction
    
    # 대체 명령어가 없는 경우 원본 명령어 반환
    print(f"No alternative found for: {instruction}")
    return instruction

# 테스트
instruction = "mov ecx, 0x20"
alternative = find_alternative(instruction)
```
